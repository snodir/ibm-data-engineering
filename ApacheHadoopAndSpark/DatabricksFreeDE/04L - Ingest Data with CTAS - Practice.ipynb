### Task 4: Compare column counts

Run the query below to compare the number of columns in `new_hires_ctas` (created with explicit column selection) versus the `employees` table (created with `SELECT *` in Lesson 1). This shows the difference between selecting specific columns and including everything.

```python
%sql
SELECT 'new_hires_ctas' AS table_name, COUNT(*) AS column_count
FROM information_schema.columns
WHERE table_schema = 'get_started_de' AND table_name = 'new_hires_ctas'
UNION ALL
SELECT 'employees', COUNT(*)
FROM information_schema.columns
WHERE table_schema = 'get_started_de' AND table_name = 'employees';
```

`new_hires_ctas` should have **4 columns** while `employees` has **5** (including `_rescued_data`). When you specify columns in your CTAS, you get exactly the schema you want.
