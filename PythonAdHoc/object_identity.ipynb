```python
_count = 0
_ls = [0]
def counter():
    global _count, _ls
    _count = _count + 1
    _ls.append(_count)
    return _count, _ls
```

```python
print(counter(), id(_count), id(_ls))
```

```python
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter
c1 = make_counter()
c2 = make_counter()
print(c1(), c1(), c1())     
print(c2(), c2())           
print(c1(), c2(), c1())     
# prints: 1 2 3
# prints: 1 2
# prints: 4 3 5
```
